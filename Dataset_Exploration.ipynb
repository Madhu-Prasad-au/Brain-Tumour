# Dataset Exploration

Inspect the supplied brain MRI dataset before training. Images are arranged in `Tumour/{train,valid,test}/{glioma,meningioma,no_tumor,pituitary}`. The data split is supplied and must be retained.


```python
from pathlib import Path
from collections import Counter
from PIL import Image
import matplotlib.pyplot as plt

data = Path("Tumour")
classes = ["glioma", "meningioma", "no_tumor", "pituitary"]
splits = ["train", "valid", "test"]
counts = {split:{c:sum(1 for p in (data/split/c).iterdir() if p.suffix.lower() in {".jpg", ".jpeg", ".png", ".bmp"}) for c in classes} for split in splits}
counts
```

```python
fig, ax = plt.subplots(figsize=(9,4))
for split in splits:
    ax.plot(classes,[counts[split][c] for c in classes],marker="o",label=split)
ax.set_ylabel("Number of images"); ax.set_title("Class counts by split"); ax.legend(); plt.xticks(rotation=20); plt.tight_layout()
```

```python
# Summarize image dimensions on a sample from every class.
for c in classes:
    files=list((data/"train"/c).glob("*.jpg"))
    sizes=Counter(Image.open(p).size for p in files[:100])
    print(c, "sampled dimensions:", sizes.most_common(5))
```

```python
# Display a randomly selected image from each class.
fig, axes=plt.subplots(1,4,figsize=(12,3))
for ax,c in zip(axes,classes):
    p=next((data/"train"/c).glob("*.jpg"))
    with Image.open(p) as im: ax.imshow(im.convert("RGB"))
    ax.set_title(c.replace("_"," ")); ax.axis("off")
plt.tight_layout()
```

## Observations

The provided split has 1,695 training, 502 validation, and 246 test images. Class counts are imbalanced, with glioma most common and no tumor least common in the training split. Source image dimensions vary; the training pipeline resizes them uniformly. Dataset labels and provenance were not independently verified.
